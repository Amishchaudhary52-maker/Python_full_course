```python
def sum(a,b):
    s=a+b
    return(s)
print(sum(2,3))
print(sum(4,5))
```

```python
def calc_avg(a,b,c):
    sum=a+b+c
    avg=sum/3
    return avg
print(calc_avg(2,3,4))
print(calc_avg(666,777,888))
```

```python
def product(a,b):
    product=a*b
    return(product)
print(product(2,3))
```

## WAP to print length of a list.(list in parameter)

```python
list=[1,2,3,4,5]
name=["amish","sudan"]
def length(list):
    length=len(list)
    return length
print(length(list))
print(length(name))
```

## WAP to print the elements of a list in a single line (list is a parameter)

```python
list=["amish","sudan","anan","alok","kusal"]
def print_list(list):
    for item in list:
        print(item,end=" ")
print(print_list(list))
```

```python

def calc_fact(n):
    fact=1
    for i in range (1,n+1):
        fact=fact*i
    return(fact)
print(calc_fact(5))
```

```python
usd_val=float(input("enter the USD:"))
def converter(USD):
    INR=USD*83
    return(INR)
print(converter(usd_val))
```

```python
def number(n):
    if(n%2==0):
        return("even")
    else:
        return("odd")
print(number(4))
```

```python
def show(n):
    if(n==0):
        return
    print(n)
    show(n-1)
print(show(5))
```

```python
def fact(n):
    if(n==1 or n==0):
        return 1
    return n*fact(n-1)
print(fact(4))
```

```python
def calc_sum(n):
    if(n==0):
        return 0
    return calc_sum(n-1)+n
print(calc_sum(5))
```

```python
def print_list(list,idx=0):
    if(idx==len(list)):
        return
    print(list[idx])
    print_list(list,idx+1)
fruits=["mango","litchi","apple","banana"]
print_list(fruits)
```
