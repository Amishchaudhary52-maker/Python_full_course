## SET in python

```python
collection={1,2,4}
print(collection)
print(type(collection))
print(len(collection))#total no. of items
student=set() #empty set
print(student)
student.add(1)
student.add("Amish")
print(student)
student.remove(1)
print(student)
collection.pop() # removes random value
print(collection)
collection.clear()
print(len(collection))
collection.add(1)
collection.add("amish")
collection.add("Chaudhary")
print(collection)
print(collection.union(student))
print(collection.intersection(student))
print(student)
print(collection)
set1={1,2,3}
set2={2,4,5}
print(set1.intersection(set2                                                                                                                                                                                                                                                                                                                                                                                                                        ))
```
