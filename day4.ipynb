## WAP to check if a list contains a palindrome elements (hint:use  copy() method)

```python
list=[1,2,3,2,1]
list1_copy=list.copy()
print(list1_copy)
list1_copy.reverse()
if list==list1_copy:
    print("It is palindrome.")
else:
    print("it is not palindrome.")
```
